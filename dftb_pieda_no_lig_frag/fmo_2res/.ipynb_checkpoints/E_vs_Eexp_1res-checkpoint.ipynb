```python
import os
import csv
import matplotlib.pyplot as plt

# === Updated Paths ===
root_dir = "/mnt/lustre/koa/lab/tsqc_group/loreab/fmo_1res"
complexes_dir = os.path.join(root_dir, "complexes_spe")
experimental_file = os.path.join(root_dir, "experimental_data.csv")

# === Load experimental data ===
experimental = {}
with open(experimental_file, newline="") as f:
    reader = csv.reader(f)
    next(reader)  # skip header
    for row in reader:
        if len(row) >= 2:
            ligand = row[0].strip()
            try:
                experimental[ligand] = float(row[1])
            except ValueError:
                continue

# === Loop over ligand subdirectories ===
eexp_list = []
eint_list = []
ligand_names = []

for ligand in os.listdir(complexes_dir):
    ligand_dir = os.path.join(complexes_dir, ligand)
    if not os.path.isdir(ligand_dir):
        continue

    avg_file = os.path.join(ligand_dir, f"{ligand}_avg_all_energies.csv")
    if not os.path.exists(avg_file):
        continue

    eint = 0.0
    with open(avg_file, newline="") as f:
        reader = csv.reader(f)
        next(reader)  # skip header
        for row in reader:
            try:
                ees = float(row[1])
                e0 = float(row[2])
                edisp = float(row[3])
                gsol = float(row[4])
                eint += ees + e0 + edisp + gsol
            except (ValueError, IndexError):
                continue

    if ligand in experimental:
        eexp_list.append(experimental[ligand])
        eint_list.append(eint)
        ligand_names.append(ligand)

# === Plotting ===
plt.figure(figsize=(8, 6))
colors = plt.colormaps['turbo'].resampled(len(ligand_names))

for i in range(len(ligand_names)):
    plt.scatter(eexp_list[i], eint_list[i], color=colors(i), s=80, edgecolor="black", label=ligand_names[i])

plt.xlabel("Experimental Binding Energy (Eexp)")
plt.ylabel("Computed Interaction Energy (Eint)")
plt.title("Eint vs Eexp across ligands (1res/frag)")
plt.grid(True)
plt.legend(loc='center left', bbox_to_anchor=(1.05, 0.5), fontsize=8, ncol=2, title="Ligands")
plt.tight_layout()
plt.savefig("Eint_vs_Eexp_all_ligands.pdf", bbox_inches='tight')
plt.show()
```

```python
import os
import csv
import matplotlib.pyplot as plt

# === Configuration ===
root_dir = "/mnt/lustre/koa/lab/tsqc_group/loreab/fmo_1res"
complexes_dir = os.path.join(root_dir, "complexes_spe")
experimental_file = os.path.join(root_dir, "experimental_data.csv")
selected_ligands = ["kb_18", "kb_04", "ad_12", "kb_01", "kb_03", "kk_98"]

# === Load experimental data ===
experimental = {}
with open(experimental_file, newline="") as f:
    reader = csv.reader(f)
    next(reader)  # skip header
    for row in reader:
        if len(row) >= 2:
            ligand = row[0].strip()
            try:
                experimental[ligand] = float(row[1])
            except ValueError:
                continue

# === Loop only over selected ligands ===
eexp_list = []
eint_list = []
ligand_names = []

for ligand in selected_ligands:
    ligand_dir = os.path.join(complexes_dir, ligand)
    if not os.path.isdir(ligand_dir):
        print(f"Directory not found for {ligand} — skipping")
        continue

    avg_file = os.path.join(ligand_dir, f"{ligand}_avg_all_energies.csv")
    if not os.path.exists(avg_file):
        print(f"File not found: {avg_file} — skipping")
        continue

    eint = 0.0
    with open(avg_file, newline="") as f:
        reader = csv.reader(f)
        next(reader)  # skip header
        for row in reader:
            try:
                ees = float(row[1])
                e0 = float(row[2])
                edisp = float(row[3])
                gsol = float(row[4])
                eint += ees + e0 + edisp + gsol
            except (ValueError, IndexError):
                continue

    if ligand in experimental:
        eexp_list.append(experimental[ligand])
        eint_list.append(eint)
        ligand_names.append(ligand)
    else:
        print(f"No experimental data for {ligand} — skipping")

# === Plotting ===
plt.figure(figsize=(8, 6))
colors = plt.colormaps['turbo'].resampled(len(ligand_names))

for i in range(len(ligand_names)):
    plt.scatter(eexp_list[i], eint_list[i], color=colors(i), s=80, edgecolor="black", label=ligand_names[i])

plt.xlabel("Experimental Binding Energy (Eexp)")
plt.ylabel("Computed Interaction Energy (Eint)")
plt.title("Eint vs Eexp for Selected Ligands")
plt.grid(True)
plt.legend(loc='center left', bbox_to_anchor=(1.05, 0.5), fontsize=8, ncol=1, title="Ligands")
plt.tight_layout()
plt.savefig("Eint_vs_Eexp_selected_ligands.pdf", bbox_inches='tight')
plt.show()
```

```python
import os
import csv
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors

# === Updated path ===
root_dir = "/mnt/lustre/koa/lab/tsqc_group/loreab/fmo_1res/complexes_spe"

# === Collect all ligand names and fragment names ===
ligands = []
fragments_set = set()
eint_data = {}

for ligand in sorted(os.listdir(root_dir)):
    subdir = os.path.join(root_dir, ligand)
    if not os.path.isdir(subdir):
        continue

    csv_file = os.path.join(subdir, f"{ligand}_avg_all_energies.csv")
    if not os.path.exists(csv_file):
        continue

    ligands.append(ligand)
    eint_data[ligand] = {}

    with open(csv_file, newline='') as f:
        reader = csv.reader(f)
        next(reader)  # skip header
        for row in reader:
            try:
                fragment = row[0].strip()
                ees = float(row[1])
                e0 = float(row[2])
                edisp = float(row[3])
                gsol = float(row[4])
                eint = ees + e0 + edisp + gsol
                eint_data[ligand][fragment] = eint
                fragments_set.add(fragment)
            except:
                continue

# === Sort fragments for consistent heatmap ordering ===
fragments = sorted(list(fragments_set))

# === Build the matrix ===
matrix = np.full((len(ligands), len(fragments)), np.nan)  # fill missing with NaN
for i, ligand in enumerate(ligands):
    for j, fragment in enumerate(fragments):
        matrix[i, j] = eint_data[ligand].get(fragment, np.nan)

# === Custom color map ===
colors = [
    (0.0, "#00FF00"),  # green for strong negative
    (0.5, "#FFFFFF"),  # white at zero
    (1.0, "#FF0000"),  # red for strong positive
]
cmap = mcolors.LinearSegmentedColormap.from_list("green_white_red", colors)

# === Plot heatmap ===
square_width = 0.633
square_height = 0.3

fig_width = square_width * len(fragments)
fig_height = square_height * len(ligands)

plt.figure(figsize=(fig_width, fig_height))

im = plt.imshow(matrix, cmap=cmap, aspect='auto', interpolation='none')

plt.xticks(ticks=np.arange(len(fragments)), labels=fragments, rotation=90)
plt.yticks(ticks=np.arange(len(ligands)), labels=ligands)

cbar = plt.colorbar(im)
cbar.set_label("Eint (kcal/mol)")

plt.title("Eint Heatmap: Ligands vs. Fragments")
plt.tight_layout()
plt.savefig("heatmap_eint_ligands_vs_fragments.png", bbox_inches='tight', dpi=300)
plt.show()
```

```python
import os
import csv
import matplotlib.pyplot as plt

# === Config ===
root_dir = "/mnt/lustre/koa/lab/tsqc_group/loreab/fmo_1res/complexes_spe"
target_fragment = "Lys-106"

ligands = []
ees_list = []
e0_list = []
edisp_list = []
gsol_list = []

for ligand in sorted(os.listdir(root_dir)):
    ligand_dir = os.path.join(root_dir, ligand)
    if not os.path.isdir(ligand_dir):
        continue

    csv_file = os.path.join(ligand_dir, f"{ligand}_avg_all_energies.csv")
    if not os.path.exists(csv_file):
        continue

    with open(csv_file, newline='') as f:
        reader = csv.reader(f)
        next(reader)  # skip header
        found = False
        for row in reader:
            if row[0].strip() == target_fragment:
                try:
                    ees = float(row[1])
                    e0 = float(row[2])
                    edisp = float(row[3])
                    gsol = float(row[4])
                except ValueError:
                    continue

                ligands.append(ligand)
                ees_list.append(ees)
                e0_list.append(e0)
                edisp_list.append(edisp)
                gsol_list.append(gsol)
                found = True
                break  # no need to read rest of file
        if not found:
            print(f"{target_fragment} not found in {ligand}")

# === Plotting ===
bar_width = 0.6
x = range(len(ligands))

plt.figure(figsize=(10, 6))
plt.bar(x, ees_list, bar_width, label='Ees', color='blue')
plt.bar(x, e0_list, bar_width, bottom=ees_list, label='E0', color='green')
bottom_2 = [ees_list[i] + e0_list[i] for i in range(len(ees_list))]
plt.bar(x, edisp_list, bar_width, bottom=bottom_2, label='Edisp', color='orange')
bottom_3 = [bottom_2[i] + edisp_list[i] for i in range(len(ees_list))]
plt.bar(x, gsol_list, bar_width, bottom=bottom_3, label='Gsol', color='red')

plt.xticks(ticks=x, labels=ligands, rotation=45, ha='right')
plt.ylabel("Eint Component Value (kcal/mol)")
plt.title(f"Energy Decomposition for {target_fragment}")
plt.legend()
plt.tight_layout()
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.savefig(f"energy_components_{target_fragment.replace('-', '_')}.pdf", dpi=300)
plt.show()
```

```python
import os
import csv
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.colors as mcolors

# === Config ===
root_dir = "/mnt/lustre/koa/lab/tsqc_group/loreab/fmo_1res/complexes_spe"
target_fragment = "Lys-106"
components = ["Ees", "E0", "Edisp", "Gsol"]

ligands = []
matrix = []

for ligand in sorted(os.listdir(root_dir)):
    ligand_dir = os.path.join(root_dir, ligand)
    if not os.path.isdir(ligand_dir):
        continue

    csv_file = os.path.join(ligand_dir, f"{ligand}_avg_all_energies.csv")
    if not os.path.exists(csv_file):
        continue

    with open(csv_file, newline='') as f:
        reader = csv.reader(f)
        next(reader)  # skip header
        for row in reader:
            if row[0].strip() == target_fragment:
                try:
                    ees = float(row[1])
                    e0 = float(row[2])
                    edisp = float(row[3])
                    gsol = float(row[4])
                    matrix.append([ees, e0, edisp, gsol])
                    ligands.append(ligand)
                except ValueError:
                    continue
                break

if not matrix:
    print("No data found for Lys-106 in any ligand.")
    exit()

matrix = np.array(matrix)

# === Color map ===
colors = [
    (0.0, "#00FF00"),  # strong negative = green
    (0.5, "#FFFFFF"),  # neutral = white
    (1.0, "#FF0000"),  # strong positive = red
]
cmap = mcolors.LinearSegmentedColormap.from_list("green_white_red", colors)

# === Plot ===
fig, ax = plt.subplots(figsize=(6, 0.5 * len(ligands)))
im = ax.imshow(matrix, cmap=cmap, aspect="auto")

# Tick labels
ax.set_xticks(np.arange(len(components)))
ax.set_xticklabels(components, rotation=45, ha='right')
ax.set_yticks(np.arange(len(ligands)))
ax.set_yticklabels(ligands)

# Colorbar
cbar = plt.colorbar(im)
cbar.set_label("Energy (kcal/mol)")

plt.title("Lys-106 Energy Components Across Ligands")
plt.tight_layout()
plt.savefig("heatmap_Lys106_components.pdf", bbox_inches='tight', dpi=300)
plt.show()
```
