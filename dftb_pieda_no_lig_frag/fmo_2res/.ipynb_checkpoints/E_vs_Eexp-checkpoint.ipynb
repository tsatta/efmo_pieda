```python
import csv
import matplotlib.pyplot as plt

# File paths
avg_ees_file = "path/to/avg_ees.csv"
experimental_file = "path/to/experimental.csv"  # e.g. contains: Ligand,E_exp

# Load avg(Ees) into a dictionary
avg_ees_dict = {}
with open(avg_ees_file, newline="") as f:
    reader = csv.reader(f)
    next(reader)
    for row in reader:
        ligand = row[0].strip()
        try:
            ees = float(row[1])
            avg_ees_dict[ligand] = ees
        except ValueError:
            continue

# Load experimental values and pair them with Ees
x_ees = []
y_exp = []

with open(experimental_file, newline="") as f:
    reader = csv.reader(f)
    next(reader)  # skip header
    for row in reader:
        ligand = row[0].strip()
        try:
            exp = float(row[1])
        except ValueError:
            continue

        if ligand in avg_ees_dict:
            x_ees.append(avg_ees_dict[ligand])
            y_exp.append(exp)

# === Plot ===
plt.figure(figsize=(8, 6))
plt.scatter(x_ees, y_exp, color='darkgreen', edgecolor='k')
plt.xlabel("Avg(Ees) [kcal/mol]")
plt.ylabel("Experimental Binding Energy [kcal/mol]")
plt.title("Avg(Ees) vs Experimental Binding Energy")
plt.grid(True)
plt.tight_layout()
plt.show()
```
